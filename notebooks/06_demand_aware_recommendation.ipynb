```python
import pandas as pd
import numpy as np
from pathlib import Path

from IPython.display import display

print("Libraries imported successfully.")
```

```python
PROJECT_DIR = Path("../")

RESULTS_DIR = PROJECT_DIR / "results"
DATA_DIR = PROJECT_DIR / "data"

print("Project directory:", PROJECT_DIR.resolve())
print("Results directory:", RESULTS_DIR.resolve())
print("Results directory exists:", RESULTS_DIR.exists())
```

```python
b5_path = RESULTS_DIR / "b5_sbert_semantic_recommendations.csv"

b5 = pd.read_csv(b5_path)

print("B5 loaded successfully.")
print("Shape:", b5.shape)

print("\nColumns:")
print(b5.columns.tolist())
```

```python
display(b5.head(10))
```

```python
print("Users:", b5["user_id"].nunique())
print("Rows:", len(b5))

print("\nRank range:")
print("Minimum rank:", b5["rank"].min())
print("Maximum rank:", b5["rank"].max())
```

```python
crowd_path = RESULTS_DIR / "crowd_predictions.csv"

crowd = pd.read_csv(crowd_path)

print("Crowd predictions loaded successfully.")
print("Shape:", crowd.shape)

print("\nColumns:")
print(crowd.columns.tolist())
```

```python
display(crowd.head(10))
```

```python
print("Number of destinations:", crowd["destination_id"].nunique())

print("\nPredicted crowd score:")
print("Minimum:", crowd["predicted_crowd_score"].min())
print("Maximum:", crowd["predicted_crowd_score"].max())
print("Mean:", crowd["predicted_crowd_score"].mean())

print("\nCrowd levels:")
print(crowd["predicted_crowd_level"].value_counts())
```

```python
required_b5_columns = [
    "user_id",
    "destination_id",
    "sbert_hybrid_score",
    "rank"
]

required_crowd_columns = [
    "destination_id",
    "predicted_crowd_score",
    "predicted_crowd_level"
]

print("Checking B5 columns...")

for col in required_b5_columns:
    print(col, ":", col in b5.columns)

print("\nChecking crowd columns...")

for col in required_crowd_columns:
    print(col, ":", col in crowd.columns)
```

```python
demand_aware = b5.merge(
    crowd[
        [
            "destination_id",
            "predicted_crowd_score",
            "predicted_crowd_level"
        ]
    ],
    on="destination_id",
    how="left"
)

print("Merge completed.")
print("Shape:", demand_aware.shape)
```

```python
print("Missing predicted crowd scores:",
      demand_aware["predicted_crowd_score"].isna().sum())

print("Missing crowd levels:",
      demand_aware["predicted_crowd_level"].isna().sum())
```

```python
crowd_min = demand_aware["predicted_crowd_score"].min()
crowd_max = demand_aware["predicted_crowd_score"].max()

if crowd_max == crowd_min:
    demand_aware["crowd_suitability"] = 1.0
else:
    demand_aware["crowd_suitability"] = (
        1
        - (
            demand_aware["predicted_crowd_score"] - crowd_min
        ) / (crowd_max - crowd_min)
    )

print("Crowd suitability calculated.")

print(
    demand_aware["crowd_suitability"].describe()
)
```

```python
demand_aware["demand_aware_score"] = (
    0.70 * demand_aware["sbert_hybrid_score"]
    + 0.30 * demand_aware["crowd_suitability"]
)

print("Demand-aware score calculated.")

display(
    demand_aware[
        [
            "user_id",
            "destination_id",
            "sbert_hybrid_score",
            "predicted_crowd_score",
            "crowd_suitability",
            "demand_aware_score"
        ]
    ].head(10)
)
```

```python
demand_aware["demand_aware_rank"] = (
    demand_aware
    .groupby("user_id")["demand_aware_score"]
    .rank(
        ascending=False,
        method="first"
    )
)

demand_aware["demand_aware_rank"] = (
    demand_aware["demand_aware_rank"].astype(int)
)

print("Demand-aware ranking completed.")
```

```python
print(
    demand_aware.groupby("user_id")["demand_aware_rank"]
    .agg(["min", "max", "count"])
    .head(10)
)
```

```python
final_recommendations = (
    demand_aware[
        demand_aware["demand_aware_rank"] <= 10
    ]
    .sort_values(
        ["user_id", "demand_aware_rank"]
    )
    .copy()
)

print("Final recommendations created.")
print("Rows:", len(final_recommendations))
print("Users:", final_recommendations["user_id"].nunique())
```

```python
def create_demand_explanation(row):
    
    crowd_level = row["predicted_crowd_level"]
    
    if crowd_level == "Low":
        crowd_text = "predicted low crowd conditions"
    elif crowd_level == "Medium":
        crowd_text = "predicted moderate crowd conditions"
    else:
        crowd_text = "predicted high crowd conditions"
    
    return (
        f"Recommended based on semantic preference matching "
        f"and {crowd_text}."
    )


final_recommendations["demand_explanation"] = (
    final_recommendations.apply(
        create_demand_explanation,
        axis=1
    )
)

display(
    final_recommendations[
        [
            "user_id",
            "destination_id",
            "predicted_crowd_level",
            "demand_aware_score",
            "demand_explanation"
        ]
    ].head(10)
)
```

```python
output_columns = [
    "user_id",
    "destination_id",
    "name",
    "category",
    "district",
    "estimated_cost_lkr",
    "duration_hours",
    "sustainability_score",
    "crowd_score",
    "predicted_crowd_score",
    "predicted_crowd_level",
    "weather_temp_c",
    "weather_rain_risk",
    "weather_suitability",
    "seasonality_score",
    "sbert_hybrid_score",
    "crowd_suitability",
    "demand_aware_score",
    "demand_aware_rank",
    "demand_explanation"
]

final_output = final_recommendations[
    [col for col in output_columns if col in final_recommendations.columns]
].copy()

print("Final output created.")
print("Shape:", final_output.shape)

display(final_output.head(10))
```

```python
print("========== FINAL RECOMMENDATION VALIDATION ==========")

print("Total rows:", len(final_output))
print("Unique users:", final_output["user_id"].nunique())
print(
    "Unique destinations:",
    final_output["destination_id"].nunique()
)

print(
    "\nRank minimum:",
    final_output["demand_aware_rank"].min()
)

print(
    "Rank maximum:",
    final_output["demand_aware_rank"].max()
)

print(
    "\nRows per user:"
)

print(
    final_output
    .groupby("user_id")
    .size()
    .value_counts()
)
```

```python
duplicate_count = final_output.duplicated(
    subset=["user_id", "destination_id"]
).sum()

print("Duplicate user-destination pairs:", duplicate_count)

assert duplicate_count == 0

print("✓ No duplicate recommendations.")
```

```python
sample_user = final_output["user_id"].iloc[0]

print("Sample user:", sample_user)

display(
    final_output[
        final_output["user_id"] == sample_user
    ][
        [
            "demand_aware_rank",
            "name",
            "category",
            "district",
            "sbert_hybrid_score",
            "predicted_crowd_score",
            "predicted_crowd_level",
            "crowd_suitability",
            "demand_aware_score"
        ]
    ].sort_values("demand_aware_rank")
)
```

```python
comparison = demand_aware[
    [
        "user_id",
        "destination_id",
        "name",
        "rank",
        "sbert_hybrid_score",
        "predicted_crowd_score",
        "predicted_crowd_level",
        "crowd_suitability",
        "demand_aware_score",
        "demand_aware_rank"
    ]
].copy()

comparison = comparison.sort_values(
    ["user_id", "demand_aware_rank"]
)

display(comparison.head(20))
```

```python
comparison["rank_change"] = (
    comparison["rank"]
    - comparison["demand_aware_rank"]
)

print("Rank change statistics:")
print(
    comparison["rank_change"].describe()
)
```

```python
changed = (
    comparison["rank_change"] != 0
).sum()

unchanged = (
    comparison["rank_change"] == 0
).sum()

print("Changed rankings:", changed)
print("Unchanged rankings:", unchanged)
print(
    "Percentage changed:",
    round(changed / len(comparison) * 100, 2),
    "%"
)
```

```python
output_path = RESULTS_DIR / "demand_aware_recommendations.csv"

final_output.to_csv(
    output_path,
    index=False
)

print("Saved successfully:")
print(output_path)
print("Exists:", output_path.exists())
```

```python
comparison_path = (
    RESULTS_DIR / "demand_aware_ranking_comparison.csv"
)

comparison.to_csv(
    comparison_path,
    index=False
)

print("Ranking comparison saved.")
print("Path:", comparison_path)
print("Exists:", comparison_path.exists())
```

```python
metrics = pd.DataFrame({
    "metric": [
        "total_recommendations",
        "unique_users",
        "unique_destinations",
        "mean_sbert_hybrid_score",
        "mean_predicted_crowd_score",
        "mean_crowd_suitability",
        "mean_demand_aware_score",
        "changed_rankings",
        "unchanged_rankings"
    ],
    "value": [
        len(final_output),
        final_output["user_id"].nunique(),
        final_output["destination_id"].nunique(),
        final_output["sbert_hybrid_score"].mean(),
        final_output["predicted_crowd_score"].mean(),
        final_output["crowd_suitability"].mean(),
        final_output["demand_aware_score"].mean(),
        changed,
        unchanged
    ]
})

display(metrics)
```

```python
metrics_path = (
    RESULTS_DIR / "demand_aware_metrics.csv"
)

metrics.to_csv(
    metrics_path,
    index=False
)

print("Metrics saved successfully.")
print("Path:", metrics_path)
```

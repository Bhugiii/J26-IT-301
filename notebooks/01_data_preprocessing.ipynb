```python
from pathlib import Path
import pandas as pd
import numpy as np

BASE_DIR = Path.cwd().parent
RAW_DIR = BASE_DIR / "data" / "raw"
PROCESSED_DIR = BASE_DIR / "data" / "processed"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Project:", BASE_DIR)
print("Raw data:", RAW_DIR)
print("Processed data:", PROCESSED_DIR)
```

```python
destinations = pd.read_csv(
    RAW_DIR / "destinations.csv"
)

users = pd.read_csv(
    RAW_DIR / "user_profiles.csv"
)

cases = pd.read_csv(
    RAW_DIR / "evaluation_cases_500.csv"
)

print("Destinations:", destinations.shape)
print("Users:", users.shape)
print("Evaluation cases:", cases.shape)
```

```python
if "destination_id" in destinations.columns:
    destinations = destinations.drop_duplicates(
        subset=["destination_id"]
    ).copy()

if "user_id" in users.columns:
    users = users.drop_duplicates(
        subset=["user_id"]
    ).copy()

if "case_id" in cases.columns:
    cases = cases.drop_duplicates(
        subset=["case_id"]
    ).copy()

print("Duplicate records removed.")
```

```python
def clean_text_columns(df):
    df = df.copy()

    for column in df.select_dtypes(include=["object", "string"]).columns:
        df[column] = (
            df[column]
            .astype("string")
            .str.strip()
        )

    return df


destinations = clean_text_columns(destinations)
users = clean_text_columns(users)
cases = clean_text_columns(cases)

print("Text cleaning completed.")
```

```python
def normalize_categories(df, columns):
    df = df.copy()

    for column in columns:
        if column in df.columns:
            df[column] = (
                df[column]
                .astype("string")
                .str.strip()
                .str.lower()
            )

    return df


destinations = normalize_categories(
    destinations,
    [
        "district",
        "category",
        "data_status"
    ]
)

users = normalize_categories(
    users,
    [
        "age_group",
        "traveller_type",
        "travel_style",
        "starting_location",
        "preferred_crowd",
        "preferred_weather",
        "data_status"
    ]
)

cases = normalize_categories(
    cases,
    [
        "travel_style",
        "starting_location",
        "preferred_crowd",
        "preferred_weather",
        "destination_category",
        "data_status"
    ]
)

print("Category normalization completed.")
```

```python
def clean_tags(value):
    if pd.isna(value):
        return pd.NA

    parts = str(value).split(",")

    parts = [
        part.strip().lower()
        for part in parts
        if part.strip()
    ]

    return ",".join(parts)


if "interest_tags" in destinations.columns:
    destinations["interest_tags"] = (
        destinations["interest_tags"]
        .apply(clean_tags)
    )

if "preferred_interests" in users.columns:
    users["preferred_interests"] = (
        users["preferred_interests"]
        .apply(clean_tags)
    )

if "preferred_interests" in cases.columns:
    cases["preferred_interests"] = (
        cases["preferred_interests"]
        .apply(clean_tags)
    )

print("Interest tags cleaned.")
```

```python
destination_numeric = [
    "latitude",
    "longitude",
    "estimated_cost_lkr",
    "duration_hours",
    "sustainability_score",
    "crowd_score",
    "seasonality_score",
    "sentiment_nature",
    "sentiment_service",
    "sentiment_value",
    "weather_temp_c",
    "weather_rain_risk",
    "weather_suitability"
]

for column in destination_numeric:
    if column in destinations.columns:
        destinations[column] = pd.to_numeric(
            destinations[column],
            errors="coerce"
        )

print("Destination numeric conversion completed.")
```

```python
user_numeric = [
    "budget_lkr_per_day",
    "trip_duration_days"
]

for column in user_numeric:
    if column in users.columns:
        users[column] = pd.to_numeric(
            users[column],
            errors="coerce"
        )

print("User numeric conversion completed.")
```

```python
case_numeric = [
    "user_budget_lkr_per_day",
    "trip_duration_days",
    "estimated_cost_lkr",
    "duration_hours",
    "distance_km",
    "preference_score",
    "budget_score",
    "sustainability_score",
    "crowd_fit_score",
    "context_score",
    "distance_score",
    "sentiment_score",
    "popularity_score",
    "overall_hybrid_score",
    "relevance_rating_1_to_5"
]

for column in case_numeric:
    if column in cases.columns:
        cases[column] = pd.to_numeric(
            cases[column],
            errors="coerce"
        )

print("Evaluation numeric conversion completed.")
```

```python
if "latitude" in destinations.columns:
    destinations.loc[
        ~destinations["latitude"].between(-90, 90),
        "latitude"
    ] = np.nan

if "longitude" in destinations.columns:
    destinations.loc[
        ~destinations["longitude"].between(-180, 180),
        "longitude"
    ] = np.nan

print("Location validation completed.")
```

```python
if "user_id" in cases.columns and "user_id" in users.columns:
    valid_users = set(users["user_id"])
    cases["valid_user"] = cases["user_id"].isin(valid_users)

if "destination_id" in cases.columns and "destination_id" in destinations.columns:
    valid_destinations = set(destinations["destination_id"])
    cases["valid_destination"] = cases["destination_id"].isin(
        valid_destinations
    )

if "valid_user" in cases.columns:
    print(
        "Invalid user references:",
        (~cases["valid_user"]).sum()
    )

if "valid_destination" in cases.columns:
    print(
        "Invalid destination references:",
        (~cases["valid_destination"]).sum()
    )

for column in ["valid_user", "valid_destination"]:
    if column in cases.columns:
        cases = cases.drop(columns=[column])

print("Reference validation completed.")
```

```python
if "interest_tags" in destinations.columns:
    destinations["num_interest_tags"] = (
        destinations["interest_tags"]
        .fillna("")
        .apply(
            lambda x: len(
                [i for i in x.split(",") if i.strip()]
            )
        )


if "preferred_interests" in users.columns:
    users["num_preferred_interests"] = (
        users["preferred_interests"]
        .fillna("")
        .apply(
            lambda x: len(
                [i for i in x.split(",") if i.strip()]
            )
        )

print("Feature creation completed.")
```

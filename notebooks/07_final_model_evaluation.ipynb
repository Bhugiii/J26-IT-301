```python
def calculate_hit_based_metrics(
    recommendations,
    ground_truth,
    rank_column,
    k_values=[1, 3, 5, 10]
):
    
    results = []
    
    # Relevant destinations for each user
    relevant_by_user = (
        ground_truth[
            ground_truth["relevant"] == 1
        ]
        .groupby("user_id")["destination_id"]
        .apply(set)
        .to_dict()
    )
    
    for k in k_values:
        
        precisions = []
        recalls = []
        hits = []
        
        for user_id, group in recommendations.groupby(
            "user_id"
        ):
            
            top_k = group[
                group[rank_column] <= k
            ].sort_values(rank_column)
            
            recommended_ids = set(
                top_k["destination_id"]
            )
            
            relevant_ids = relevant_by_user.get(
                user_id,
                set()
            )
            
            intersection = (
                recommended_ids
                & relevant_ids
            )
            
            hit_count = len(intersection)
            
            # Precision@K
            precision = (
                hit_count / k
                if k > 0
                else 0
            )
            
            # Recall@K
            if len(relevant_ids) > 0:
                recall = (
                    hit_count
                    / len(relevant_ids)
                )
            else:
                recall = np.nan
            
            precisions.append(precision)
            recalls.append(recall)
            hits.append(hit_count)
        
        results.append({
            "k": k,
            "precision": np.mean(precisions),
            "recall": np.nanmean(recalls),
            "average_hits": np.mean(hits)
        })
    
    return pd.DataFrame(results)
```

```python
def calculate_ndcg_from_ground_truth(
    recommendations,
    ground_truth,
    rank_column,
    k
):
    
    merged = recommendations.merge(
        ground_truth[
            [
                "user_id",
                "destination_id",
                "relevance_rating_1_to_5"
            ]
        ],
        on=[
            "user_id",
            "destination_id"
        ],
        how="left"
    )
    
    merged["relevance_rating_1_to_5"] = (
        merged["relevance_rating_1_to_5"]
        .fillna(0)
    )
    
    scores = []
    
    for user_id, group in merged.groupby(
        "user_id"
    ):
        
        group = group.sort_values(
            rank_column
        )
        
        top_k = group.head(k)
        
        relevance = (
            top_k[
                "relevance_rating_1_to_5"
            ]
            .values
        )
        
        if len(relevance) == 0:
            continue
        
        dcg = 0.0
        
        for i, rel in enumerate(
            relevance,
            start=1
        ):
            dcg += (
                (2 ** rel - 1)
                / np.log2(i + 1)
            )
        
        ideal_relevance = np.sort(
            relevance
        )[::-1]
        
        idcg = 0.0
        
        for i, rel in enumerate(
            ideal_relevance,
            start=1
        ):
            idcg += (
                (2 ** rel - 1)
                / np.log2(i + 1)
            )
        
        if idcg == 0:
            score = 0.0
        else:
            score = dcg / idcg
        
        scores.append(score)
    
    if len(scores) == 0:
        return np.nan
    
    return np.mean(scores)
```

```python
ndcg_comparison = []

for k in [1, 3, 5, 10]:
    
    demand_ndcg = calculate_ndcg_from_ground_truth(
        final_recommendations,
        ground_truth,
        "demand_aware_rank",
        k
    )
    
    b5_ndcg = calculate_ndcg_from_ground_truth(
        b5_eval_clean,
        ground_truth,
        "rank",
        k
    )
    
    ndcg_comparison.append({
        "k": k,
        "B5_NDCG": b5_ndcg,
        "Demand_Aware_NDCG": demand_ndcg,
        "NDCG_change":
            demand_ndcg - b5_ndcg
    })

ndcg_comparison = pd.DataFrame(
    ndcg_comparison
)

display(ndcg_comparison)
```
